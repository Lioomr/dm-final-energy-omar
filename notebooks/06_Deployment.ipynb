This notebook documents the deployment layer of the project: the Streamlit dashboard. The dashboard is not a production cloud deployment, but it is the interactive delivery layer required by the project brief.

````python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DASHBOARD_PATH = PROJECT_ROOT / "dashboard" / "app.py"
PREPARED_PATH = PROJECT_ROOT / "outputs" / "prepared_hourly_energy.csv"

print("Dashboard file exists:", DASHBOARD_PATH.exists())
print("Prepared data exists:", PREPARED_PATH.exists())
print("Dashboard path:", DASHBOARD_PATH)
````

## Dashboard Purpose

The Streamlit dashboard presents the project results in an interactive format. It allows a user to filter by date range and inspect the major data mining outputs without reading every notebook cell.

## Dashboard Sections

- **Overview:** high-level consumption metrics and time-series visualizations.
- **Clustering:** behavior segments, silhouette metrics, cluster profile table, and PCA cluster projection.
- **Regression:** model comparison using MAE, RMSE, and R2.
- **Classification:** high-vs-normal consumption model accuracy and classification report.
- **Anomaly Detection:** unusual consumption periods and top anomaly table.
- **Business Insights:** peak hour, high-consumption share, anomaly share, and practical recommendations.

## How to Run the Dashboard

Run this command from the project root:

```bash
streamlit run dashboard/app.py
```

Then open:

```text
http://localhost:8501
```

## Deployment Notes

## Final Business Recommendations

- Monitor evening peak hours, especially around the highest average consumption hour.
- Shift heavy appliance use away from peak periods where possible.
- Investigate repeated anomaly periods, especially high-power events with high current intensity.
- Use high-consumption classification as a basis for simple household energy alerts.
- Extend the project later with weather, occupancy, or tariff data for stronger recommendations.
