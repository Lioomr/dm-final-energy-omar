# 01 - Business Understanding

This notebook defines the business problem, project goals, research questions, and success criteria for the Smart Energy Consumption Analytics System.

The project follows the CRISP-DM methodology. This first phase explains why household energy consumption is worth analyzing and how data mining can support better decisions.

## Business Problem

Households often receive electricity bills without understanding when consumption peaks, which usage patterns repeat, or which periods look abnormal. This lack of visibility makes it difficult to reduce waste or shift heavy appliance usage away from peak hours.

A smart energy analytics workflow can help transform raw meter readings into practical insights: peak-hour detection, consumption prediction, high-usage classification, and anomaly detection.

## Project Objectives

- Analyze household electricity consumption patterns over time.
- Identify peak usage hours and repeated behavior patterns.
- Predict short-term global active power usage.
- Classify periods as high or normal consumption.
- Detect unusual consumption spikes that may require investigation.
- Present the results in an interactive Streamlit dashboard.

## Research Questions

1. When is household electricity usage highest?
2. Can similar consumption behaviors be grouped using clustering?
3. Can recent consumption history help predict short-term power usage?
4. Can high-consumption periods be classified accurately?
5. Which periods appear unusual compared with normal household behavior?

## Data Mining Techniques Selected

- **Clustering:** K-Means and DBSCAN for grouping similar usage behavior.
- **Regression:** Linear Regression, Ridge Regression, and Polynomial Ridge for short-term consumption prediction.
- **Classification:** Logistic Regression and Random Forest for high-vs-normal consumption labels.
- **Anomaly Detection:** Isolation Forest and PCA-based analysis for unusual usage periods.

## Success Criteria

The project is successful if it produces a reproducible notebook workflow, meaningful visualizations, baseline model results, interpretable energy insights, and a dashboard that communicates the results clearly.

The goal is not perfect accuracy. The goal is to demonstrate a complete data mining process from business understanding through deployment.
